```python

# Hospital Ward Staffing Optimizer (A1)
# Method: System of Linear Equations + Gauss Elimination

# --------------------------------------------------
# INPUT DATA
# --------------------------------------------------

# Staffing coefficients:
# Columns = Senior, Junior, Assistant
# Rows = Ward A, Ward B, Ward C
#
# IMPORTANT:
# These are sample values because the original problem
# does not specify the 9 staffing coefficients.

A = [
    [2, 1, 1],   # Ward A
    [1, 2, 1],   # Ward B
    [1, 1, 2]    # Ward C
]

# Required staff-hours
B = [20, 25, 15]

# Cost per hour
cost = [800, 400, 200]

# Staff names
staff_names = ["Senior Nurse", "Junior Nurse", "Assistant"]
wards = ["Ward A", "Ward B", "Ward C"]


# --------------------------------------------------
# GAUSS ELIMINATION
# --------------------------------------------------

def gauss_elimination(A, B):

    n = len(B)

    # Create augmented matrix [A | B]
    matrix = [
        A[i][:] + [B[i]]
        for i in range(n)
    ]

    print("\nAugmented Matrix [A | B]:")
    for row in matrix:
        print(row)

    # Forward elimination
    for i in range(n):

        # Find a suitable pivot
        max_row = i
        for k in range(i + 1, n):
            if abs(matrix[k][i]) > abs(matrix[max_row][i]):
                max_row = k

        # Swap rows
        matrix[i], matrix[max_row] = matrix[max_row], matrix[i]

        if abs(matrix[i][i]) < 1e-10:
            raise ValueError("No unique solution exists.")

        # Eliminate values below pivot
        for k in range(i + 1, n):
            factor = matrix[k][i] / matrix[i][i]

            for j in range(i, n + 1):
                matrix[k][j] -= factor * matrix[i][j]

        print(f"\nAfter elimination step {i + 1}:")
        for row in matrix:
            print([round(value, 3) for value in row])

    # Back substitution
    x = [0.0] * n

    for i in range(n - 1, -1, -1):
        total = matrix[i][n]

        for j in range(i + 1, n):
            total -= matrix[i][j] * x[j]

        x[i] = total / matrix[i][i]

    return x


# --------------------------------------------------
# SOLVE SYSTEM
# --------------------------------------------------

solution = gauss_elimination(A, B)


# --------------------------------------------------
# DISPLAY STAFFING SOLUTION
# --------------------------------------------------

print("\n" + "=" * 45)
print("HOSPITAL WARD STAFFING OPTIMIZER")
print("=" * 45)

print("\nOptimal/Calculated Staffing Mix:")

for i in range(len(solution)):
    print(f"{staff_names[i]:15}: {solution[i]:.2f} hours")


# --------------------------------------------------
# CALCULATE TOTAL COST
# --------------------------------------------------

total_cost = sum(
    solution[i] * cost[i]
    for i in range(len(solution))
)

print("\nStaffing Cost:")
for i in range(len(solution)):
    individual_cost = solution[i] * cost[i]
    print(
        f"{staff_names[i]:15}: "
        f"₹{individual_cost:.2f}"
    )

print("-" * 45)
print(f"TOTAL COST      : ₹{total_cost:.2f}")


# --------------------------------------------------
# VERIFY WARD REQUIREMENTS
# --------------------------------------------------

print("\nWard Requirement Verification:")

for i in range(len(A)):

    calculated_hours = sum(
        A[i][j] * solution[j]
        for j in range(len(solution))
    )

    print(
        f"{wards[i]}: "
        f"Required = {B[i]:.2f}, "
        f"Provided = {calculated_hours:.2f}"
    )


print("\n✓ Gauss elimination completed successfully.")
print("✓ Staffing requirements verified.")
```
