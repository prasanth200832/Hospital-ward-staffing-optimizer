```python

# ============================================================
# HOSPITAL WARD STAFFING OPTIMIZER (A1)
# B.Tech Mathematics Mini Project
# Mathematical Concept: AX = B + Gaussian Elimination
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 1. INPUTS
# ------------------------------------------------------------

wards = ["Ward A", "Ward B", "Ward C"]
staff = ["Senior Nurses", "Junior Nurses", "Assistants"]

# Required staff-hours for each ward
B = np.array([20, 25, 15], dtype=float)

# Cost per hour
cost = np.array([800, 400, 200], dtype=float)

# Allocation matrix A
#
#             Senior   Junior   Assistant
# Ward A        0.50     0.30       0.20
# Ward B        0.30     0.50       0.20
# Ward C        0.20     0.20       0.60
#
# Each column represents one staff type.

A = np.array([
    [0.50, 0.30, 0.20],
    [0.30, 0.50, 0.20],
    [0.20, 0.20, 0.60]
], dtype=float)


# ------------------------------------------------------------
# 2. GAUSSIAN ELIMINATION
# ------------------------------------------------------------

def gaussian_elimination(A, B):

    # Create augmented matrix [A | B]
    M = np.column_stack((A.astype(float), B.astype(float)))

    n = len(B)

    print("\nInitial Augmented Matrix [A | B]:")
    print(np.round(M, 4))

    # Forward elimination
    for col in range(n):

        # Partial pivoting
        pivot = col + np.argmax(abs(M[col:, col]))

        if abs(M[pivot, col]) < 1e-12:
            raise ValueError("No unique solution.")

        # Swap rows if required
        if pivot != col:
            M[[col, pivot]] = M[[pivot, col]]

        # Eliminate values below pivot
        for row in range(col + 1, n):

            factor = M[row, col] / M[col, col]

            M[row] = M[row] - factor * M[col]

    # Back substitution
    X = np.zeros(n)

    for row in range(n - 1, -1, -1):

        X[row] = (
            M[row, -1]
            - np.dot(M[row, row + 1:n], X[row + 1:n])
        ) / M[row, row]

    return X


# ------------------------------------------------------------
# 3. SOLVE AX = B
# ------------------------------------------------------------

X = gaussian_elimination(A, B)

# ------------------------------------------------------------
# 4. CALCULATE COST
# ------------------------------------------------------------

staff_cost = X * cost
total_cost = np.sum(staff_cost)

# Verify solution
calculated_B = A @ X


# ------------------------------------------------------------
# 5. OUTPUT
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("HOSPITAL WARD STAFFING OPTIMIZER")
print("=" * 60)

print("\nMathematical Model:")
print("AX = B")

print("\nMatrix A:")
print(A)

print("\nRequired Hours B:")
print(B)

print("\nSolution X:")

for i in range(3):
    print(
        f"{staff[i]:20s} = {X[i]:.2f} hours"
    )

print("\nStaffing Cost:")

for i in range(3):
    print(
        f"{staff[i]:20s} : "
        f"{X[i]:.2f} × ₹{cost[i]:.0f} "
        f"= ₹{staff_cost[i]:,.2f}"
    )

print("-" * 60)
print(f"TOTAL STAFFING COST = ₹{total_cost:,.2f}")

# ------------------------------------------------------------
# 6. VERIFICATION
# ------------------------------------------------------------

print("\nVerification of AX = B:")

verification = pd.DataFrame({
    "Ward": wards,
    "Required Hours": B,
    "Calculated Hours": calculated_B,
    "Difference": calculated_B - B
})

display(verification)

if np.allclose(calculated_B, B):
    print("✓ Solution verified successfully.")
else:
    print("✗ Verification failed.")


# ------------------------------------------------------------
# 7. FINAL RESULT TABLE
# ------------------------------------------------------------

result = pd.DataFrame({
    "Staff Type": staff,
    "Hours": np.round(X, 2),
    "Cost/Hour (₹)": cost.astype(int),
    "Total Cost (₹)": np.round(staff_cost, 2)
})

print("\nFINAL STAFFING PLAN")
display(result)

print(f"\nTOTAL COST: ₹{total_cost:,.2f}")
```
